## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Bagaimana pola rata-rata jumlah penyewaan sepeda (cnt) berdasarkan jam dalam sehari selama periode 2011–2012, dan pada jam berapa permintaan penyewaan mencapai puncaknya?
- **Pertanyaan 2:** Bagaimana perbedaan rata-rata jumlah penyewaan sepeda (cnt) berdasarkan kondisi cuaca (weathersit) selama periode 2011–2012, dan kondisi cuaca mana yang memiliki tingkat penyewaan tertinggi?

### Cleaning Data

#### Fixing ... problem

```python

# Mengubah tipe data dteday menjadi datetime
hour_df['dteday'] = pd.to_datetime(hour_df['dteday'])
day_df['dteday'] = pd.to_datetime(day_df['dteday'])

# Mengecek hasil perubahan
print("Tipe data dteday Hourly:",
      hour_df['dteday'].dtype)
print("Tipe data dteday Daily:",
      day_df['dteday'].dtype)
```

```python

# Daftar kolom kategorikal
categorical_columns = [
    'season',
    'yr',
    'mnth',
    'holiday',
    'weekday',
    'workingday',
    'weathersit'
]

# Mengubah tipe data kolom kategorikal
for col in categorical_columns:
    hour_df[col] = hour_df[col].astype('category')
    day_df[col] = day_df[col].astype('category')

# Khusus dataset hourly, kolom hr juga kategorikal
hour_df['hr'] = hour_df['hr'].astype('category')
print("Tipe Data Hourly Dataset:")
print(hour_df.dtypes)
print("\nTipe Data Daily Dataset:")
print(day_df.dtypes)
```

```python

# Mengecek kembali informasi dataset
print("Hourly Dataset:")
hour_df.info()
print("\nDaily Dataset:")
day_df.info()

# Mengecek kembali missing values
print("\nMissing Values:")
print("Hourly:", hour_df.isna().sum().sum())
print("Daily:", day_df.isna().sum().sum())

# Mengecek kembali duplicate data
print("\nDuplicate Data:")
print("Hourly:", hour_df.duplicated().sum())
print("Daily:", day_df.duplicated().sum())
```


**Insight:** (Opsional)
- Kolom dteday pada kedua dataset berhasil diubah menjadi tipe data datetime.
- Kolom season, yr, mnth, holiday, weekday, workingday, dan weathersit berhasil diubah menjadi tipe data category.
- Kolom hr pada hourly dataset juga diubah menjadi category karena menunjukkan kelompok jam penyewaan.
- Tidak dilakukan penghapusan baris maupun pengisian nilai kosong karena dataset tidak memiliki missing values dan duplicate data.
- Setelah proses cleaning, kedua dataset siap digunakan untuk tahap Exploratory Data Analysis (EDA).


## Exploratory Data Analysis (EDA)

### Explore Jumlah Penyewaan Sepeda Berdasarkan Jam


```python
# Menghitung rata-rata dan total penyewaan berdasarkan jam
hourly_rentals = hour_df.groupby('hr', observed=True)['cnt'].agg(
    average_rentals='mean',
    total_rentals='sum'
).reset_index()

# Mengurutkan berdasarkan jam
hourly_rentals = hourly_rentals.sort_values('hr')

# Menampilkan hasil
display(hourly_rentals.round(2))
```

```python

# Mencari jam dengan rata-rata penyewaan tertinggi
peak_hour = hourly_rentals.loc[
    hourly_rentals['average_rentals'].idxmax()
]

# Mencari jam dengan rata-rata penyewaan terendah
lowest_hour = hourly_rentals.loc[
    hourly_rentals['average_rentals'].idxmin()
]

print("Jam dengan penyewaan tertinggi:")
print(f"Jam {int(peak_hour['hr'])}:00 dengan rata-rata "
      f"{peak_hour['average_rentals']:.2f} sepeda")
print("\nJam dengan penyewaan terendah:")
print(f"Jam {int(lowest_hour['hr'])}:00 dengan rata-rata "
      f"{lowest_hour['average_rentals']:.2f} sepeda")
```

### Explore Jumlah Penyewaan Sepeda Berdasarkan Kondisi Cuaca


```python

# Menghitung rata-rata dan total penyewaan berdasarkan cuaca
weather_rentals = day_df.groupby('weathersit', observed=True)['cnt'].agg(
    average_rentals='mean',
    total_rentals='sum',
    total_days='count'
).reset_index()
# Mengurutkan berdasarkan rata-rata penyewaan
weather_rentals = weather_rentals.sort_values(
    'average_rentals', ascending=False
)
# Menampilkan hasil
display(weather_rentals.round(2))
```

```python

# Mencari cuaca dengan rata-rata penyewaan tertinggi
best_weather = weather_rentals.loc[
    weather_rentals['average_rentals'].idxmax()
]

# Mencari cuaca dengan rata-rata penyewaan terendah
worst_weather = weather_rentals.loc[
    weather_rentals['average_rentals'].idxmin()
]

print("Kondisi cuaca dengan penyewaan tertinggi:")
print(f"Weathersit {int(best_weather['weathersit'])} "
      f"dengan rata-rata {best_weather['average_rentals']:.2f} sepeda")
print("\nKondisi cuaca dengan penyewaan terendah:")
print(f"Weathersit {int(worst_weather['weathersit'])} "
      f"dengan rata-rata {worst_weather['average_rentals']:.2f} sepeda")
```

**Insight:** (Opsional)
- Berdasarkan hasil pengelompokan penyewaan berdasarkan jam, rata-rata penyewaan sepeda tertinggi terjadi pada pukul 17.00 dengan rata-rata 461,45 penyewaan.
- Rata-rata penyewaan terendah terjadi pada pukul 04.00 dengan rata-rata 6,35 penyewaan.
- Aktivitas penyewaan cenderung meningkat pada pagi hari, khususnya pukul 07.00–08.00, dan kembali meningkat pada sore hari.
- Berdasarkan kondisi cuaca, cuaca cerah (weathersit 1) memiliki rata-rata penyewaan harian tertinggi sebesar 4.876,79.
- Kondisi cuaca dengan hujan ringan atau salju ringan (weathersit 3) memiliki rata-rata penyewaan harian terendah sebesar 1.803,29.
- Hasil analisis menunjukkan bahwa waktu dan kondisi cuaca memiliki pola perbedaan jumlah penyewaan sepeda yang dapat menjadi pertimbangan dalam perencanaan operasional.


## Visualization & Explanatory Analysis

### Pertanyaan 1:
Bagaimana pola rata-rata jumlah penyewaan sepeda (`cnt`) berdasarkan jam dalam sehari selama periode 2011–2012, dan pada jam berapa permintaan penyewaan mencapai puncaknya?

```python

# Visualisasi rata-rata penyewaan berdasarkan jam

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=hourly_rentals,
    x='hr',
    y='average_rentals',
    marker='o'
)

plt.title('Rata-rata Penyewaan Sepeda Berdasarkan Jam')
plt.xlabel('Jam')
plt.ylabel('Rata-rata Jumlah Penyewaan')
plt.xticks(range(24))
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Pertanyaan 2:
Bagaimana perbedaan rata-rata jumlah penyewaan sepeda (`cnt`) berdasarkan kondisi cuaca (`weathersit`) selama periode 2011–2012, dan kondisi cuaca mana yang memiliki tingkat penyewaan tertinggi?
an 2:

```python

# Visualisasi rata-rata penyewaan berdasarkan kondisi cuaca
weather_labels = {
    1: 'Cerah / Sedikit Berawan',
    2: 'Berkabut / Berawan',
    3: 'Hujan Ringan / Salju Ringan'
}

weather_rentals['weather_name'] = weather_rentals['weathersit'].map(
    weather_labels
)

plt.figure(figsize=(10, 5))
sns.barplot(
    data=weather_rentals,
    x='weather_name',
    y='average_rentals'
)
plt.title('Rata-rata Penyewaan Sepeda Berdasarkan Kondisi Cuaca')
plt.xlabel('Kondisi Cuaca')
plt.ylabel('Rata-rata Jumlah Penyewaan')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
- Berdasarkan line chart, jumlah penyewaan sepeda mengalami peningkatan pada pagi hari dan mencapai puncak pada pukul 17.00 dengan rata-rata 461,45 penyewaan.
- Setelah pukul 17.00, jumlah penyewaan cenderung menurun hingga malam hari.
- Berdasarkan bar chart, kondisi cuaca cerah memiliki rata-rata penyewaan tertinggi sebesar 4.876,79 penyewaan per hari.
- Kondisi cuaca hujan ringan atau salju ringan memiliki rata-rata penyewaan terendah sebesar 1.803,29 penyewaan per hari.
- Kedua visualisasi menunjukkan bahwa waktu dan kondisi cuaca berkaitan dengan perbedaan tingkat permintaan penyewaan sepeda.



## Analisis Lanjutan: Binning dan Manual Grouping
Pada analisis lanjutan ini, dilakukan pengelompokan waktu penyewaan sepeda menggunakan teknik Binning. Pengelompokan dilakukan menjadi empat kategori, yaitu dini hari, pagi, siang, dan sore hingga malam.


```python

# Membuat kategori waktu berdasarkan jam
hour_df['time_category'] = pd.cut(
    hour_df['hr'].astype(int),
    bins=[-1, 5, 10, 15, 23],
    labels=['Dini Hari', 'Pagi', 'Siang', 'Sore-Malam']
)

# Menghitung rata-rata penyewaan berdasarkan kategori waktu
time_analysis = hour_df.groupby(
    'time_category', observed=True
)['cnt'].agg(
    average_rentals='mean',
    total_rentals='sum'
).reset_index()

# Menampilkan hasil
display(time_analysis.round(2))
```

```python

# Mengurutkan kategori berdasarkan rata-rata penyewaan
time_analysis_sorted = time_analysis.sort_values(
    'average_rentals', ascending=False
)

display(time_analysis_sorted.round(2))

print("Kategori waktu dengan penyewaan tertinggi:",
      time_analysis_sorted.iloc[0]['time_category'])
print("Rata-rata penyewaan:",
      round(time_analysis_sorted.iloc[0]['average_rentals'], 2))
```

**Insight:** (Opsional)
- Pengelompokan waktu dilakukan menjadi empat kategori, yaitu dini hari, pagi, siang, dan sore-malam.
- Setiap kategori memiliki karakteristik jumlah penyewaan yang berbeda.
- Hasil pengelompokan dapat digunakan untuk mengidentifikasi periode dengan permintaan penyewaan tertinggi.
- Informasi ini dapat membantu pengelola dalam menentukan alokasi sepeda dan kebutuhan operasional berdasarkan waktu.
